```python
w = torch.tensor(0.5, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
x = torch.tensor(2.0)
y_true = torch.tensor(5.0)

y_pred = w * x + b
loss = (y_pred - y_true) ** 2   # Squared error

loss.backward()
print(w.grad)   # how much changing w will affect the loss
print(b.grad)   # how much changing b will affect the loss
#this happens in any nueral network but you make this on billions of parameters 
```

```python
"""
how i can built Neural Network Model in pytorch
every neural network model in pytorch is a class that inherits from nn.Module:
__init__=> method is used to define the layers of the model
forward => method is used to define the forward pass of the model
let me give you an example
"""
import torch
import torch.nn as nn 

class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(1, 10) # one input and 10 output
        self.layer2 = nn.Linear(10, 32) # 10 inputs and 32 outputs
        self.layer3 = nn.Linear(32, 16) # 32 inputs and 16 outputs
        self.layer4 = nn.Linear(16, 1) # 16 inputs and 1 output
        self.relu = nn.ReLU() # activation function

    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        x = torch.relu(x)
        x = self.layer3(x)
        x = torch.relu(x)
        x = self.layer4(x)
        return x
model = NeuralNetwork()

# Input
x = torch.tensor([[5.0]])

# True value
y_true = torch.tensor([[0.04]])

# Loss
loss_fn = nn.MSELoss()

# Forward Pass
predictions = model(x)

# Calculate Loss
loss = loss_fn(predictions, y_true)

print("Prediction:", predictions)
print("True value:", y_true)
print("Loss:", loss.item())
```
